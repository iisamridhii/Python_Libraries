### Make a new column for Episode Count

```python
df.loc[1]
```

```python
#every 'Title' has there episode counts but it does not have a specific column
```

```python
def extract_episodes(txt):
    check = False
    data =""
    
    for ch in txt:
        if ch == "(":
           check= True
           continue
            
        if ch == ")":
            return data
            
        if check:
            data += ch
```

```python
df["Title"].apply(extract_episodes)
```

```python
df["Episodes"]= df["Title"].apply(extract_episodes)
```

# Make a new Column for Time Stamp

```python
df.loc[0]['Title']
```

```python
def  extraction_time(txt):
    check= False
    data= ""
    for i in range(len(txt)):
        if txt[i] ==')':
            for j in range(i,i+18):
                data+=txt[j]

            return data
        
    
```

```python
df['Title'].apply(extraction_time)
```

```python
df['Total time']=df['Title'].apply(extraction_time)
```
