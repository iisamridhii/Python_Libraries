### Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.subplot(2,2,4) 
plt.subplot(2,2,3) 
plt.subplot(2,2,2)
```

```python
plt.subplot(2,2,1) 
plt.plot(x,y)
plt.subplot(2,2,2) 
plt.plot(x,y)
plt.subplot(2,2,3) 
plt.plot(y,x) 
plt.subplot(2,2,3) 
plt.plot(y,x)
```

### Object Oriented Ways

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,4,8)
x
y=x**2 
y
fig=plt.figure()
ax1= fig.add_axes([0.1,0.1,1,1])
ax1.plot(x,y)
plt.show()
```

```python
fig=plt.figure()
ax1= fig.add_axes([0.1,0.1,1,1])
ax1.plot(x,y)
plt.show()
```

```python
fig=plt.figure()
ax1= fig.add_axes([0.1,0.1,1,1])
ax1.plot(x,y)
ax2=fig.add_axes([0.1,0.1,0.8,0.8])
ax2.plot(x,y)
```

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
